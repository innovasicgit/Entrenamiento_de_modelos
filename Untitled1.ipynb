```python
import pandas as pd

carpeta_base = 'C:/Users/Usuario/Downloads/Notebooks de Alex/Datasets_Entrenamiento/'

# Dataset original (ya con label y tipo_ataque)
df_botiot = pd.read_csv(carpeta_base + 'Bot-IoT.csv')

# Tus capturas nuevas (crudas, sin label ni tipo_ataque)
udp_1 = pd.read_csv(carpeta_base + 'UDP_Flood_1.csv')
udp_2 = pd.read_csv(carpeta_base + 'UDP_Flood_2.csv')
icmp_1 = pd.read_csv(carpeta_base + 'ICMP_AVANZADO_1.csv')
icmp_2 = pd.read_csv(carpeta_base + 'ICMP_AVANZADO_2.csv')

print(df_botiot.shape, udp_1.shape, udp_2.shape, icmp_1.shape, icmp_2.shape)
```

```python
for df in [udp_1, udp_2]:
    df['label'] = 1
    df['tipo_ataque'] = 'UDP_Flood'

for df in [icmp_1, icmp_2]:
    df['label'] = 1
    df['tipo_ataque'] = 'ICMP_Flood'

df_udp = pd.concat([udp_1, udp_2], ignore_index=True)
df_icmp = pd.concat([icmp_1, icmp_2], ignore_index=True)

print(df_udp.shape, df_icmp.shape)  # deberían ser ~6000 filas cada uno
```

```python
print(set(df_botiot.columns) == set(df_udp.columns))
print(set(df_botiot.columns) == set(df_icmp.columns))
```

```python
df_entrenamiento_completo = pd.concat([df_botiot, df_udp, df_icmp], ignore_index=True)
print(df_entrenamiento_completo['tipo_ataque'].value_counts())
```

```python
print(df_entrenamiento_completo['epoch_time'].isna().sum())  # debería dar ~320000
print(df_entrenamiento_completo.shape)
```

```python
df_udp_balanced = pd.concat([df_udp, df_udp], ignore_index=True)
df_icmp_balanced = pd.concat([df_icmp, df_icmp], ignore_index=True)

df_entrenamiento_completo = pd.concat(
    [df_botiot, df_udp_balanced, df_icmp_balanced],
    ignore_index=True
)
print(df_entrenamiento_completo['tipo_ataque'].value_counts())
```

```python
from sklearn.model_selection import train_test_split

df_normal_reducido, _ = train_test_split(
    df_entrenamiento_completo[df_entrenamiento_completo['tipo_ataque'] == 'Normal'],
    train_size=30000,
    random_state=42
)
df_resto = df_entrenamiento_completo[df_entrenamiento_completo['tipo_ataque'] != 'Normal']

df_entrenamiento_ocsvm = pd.concat([df_normal_reducido, df_resto], ignore_index=True)
print(df_entrenamiento_ocsvm['tipo_ataque'].value_counts())
print(df_entrenamiento_ocsvm.shape)
```

```python
df_entrenamiento_completo.to_csv(
    'C:/Users/Usuario/Downloads/Notebooks de Alex/Datasets_Entrenamiento/dataset_entrenamiento_completo.csv',
    index=False
)
print("Guardado correctamente")
```
